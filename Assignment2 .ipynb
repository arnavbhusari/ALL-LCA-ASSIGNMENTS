```python
a=float(input("Enter first no."))
b=float(input("Enter second no."))
c=float(input("Enter third no."))

if a==b and b==c:
    print("All three are equal",a)
elif a>=b and a>=c:
    print("First is the greatest no.",a)
elif b>=c and b>=a:
    print("Second is the greatest no.",b)
else :
    print("third is the greatest no.",c)
```
